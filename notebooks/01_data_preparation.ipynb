## 4. Missing-Value Analysis
Real-world healthcare datasets frequently utilize placeholder characters such as `'?'` or `'Unknown/Invalid'` to indicate missing measurements. In this section, we replace implicit missing indicators with standard `np.nan` values and quantify missingness across all features.


```python
df_proc = df_raw.copy()

missing_placeholder_cols = ['race', 'weight', 'payer_code', 'medical_specialty', 'diag_1', 'diag_2', 'diag_3']
for col in missing_placeholder_cols:
    df_proc[col] = df_proc[col].replace('?', np.nan)

df_proc['gender'] = df_proc['gender'].replace('Unknown/Invalid', np.nan)
```

```python
# Lab test results 'None' means 'Not Tested'
df_proc['max_glu_serum'] = df_proc['max_glu_serum'].fillna('Not Tested').replace('None', 'Not Tested')
df_proc['A1Cresult'] = df_proc['A1Cresult'].fillna('Not Tested').replace('None', 'Not Tested')
```

```python
missing_counts = df_proc.isnull().sum()
missing_percentages = (missing_counts / len(df_proc)) * 100

missing_df = pd.DataFrame({
    'Feature': missing_counts.index,
    'Missing_Count': missing_counts.values,
    'Missing_Percentage': missing_percentages.values
})

missing_df = missing_df[missing_df['Missing_Count'] > 0].sort_values(by='Missing_Count', ascending=False).reset_index(drop=True)

print("=== FEATURES CONTAINING MISSING VALUES ===")
missing_df
```

```python
plt.figure(figsize=(10, 5))
barplot = sns.barplot(
    data=missing_df,
    x='Missing_Percentage',
    y='Feature',
    palette='Reds_r'
)
plt.title('Missing Value Percentage by Clinical Feature', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Missing Percentage (%)', fontsize=12)
plt.ylabel('Clinical Feature', fontsize=12)

for p in barplot.patches:
    width = p.get_width()
    plt.text(width + 1.2, p.get_y() + p.get_height()/2, f'{width:.2f}%', 
             ha='left', va='center', fontsize=10, fontweight='bold')

plt.xlim(0, 110)
plt.tight_layout()
plt.show()
```

## 6. Data-Type Handling
We map relational lookup tables from `IDS_mapping.csv` to replace numeric IDs with descriptive clinical text, and drop zero-variance features.

```python
lines = ids_raw_text.split('\n')
mapping_dicts = {'admission_type_id': {}, 'discharge_disposition_id': {}, 'admission_source_id': {}}
current_key = None

for line in lines:
    line_str = line.strip()
    if not line_str or line_str == ',':
        continue
    if line_str.startswith('admission_type_id'):
        current_key = 'admission_type_id'
        continue
    elif line_str.startswith('discharge_disposition_id'):
        current_key = 'discharge_disposition_id'
        continue
    elif line_str.startswith('admission_source_id'):
        current_key = 'admission_source_id'
        continue
        
    parts = line_str.split(',', 1)
    if len(parts) == 2 and parts[0].isdigit():
        v_id = int(parts[0])
        desc = parts[1].strip('"').strip()
        if desc.upper() in ['NULL', 'NAN', 'NONE', '']:
            desc = 'Not Available / Unmapped'
        mapping_dicts[current_key][v_id] = desc

df_proc['admission_type_desc'] = df_proc['admission_type_id'].map(mapping_dicts['admission_type_id']).fillna('Not Available / Unmapped').replace('NULL', 'Not Available / Unmapped')
df_proc['discharge_disposition_desc'] = df_proc['discharge_disposition_id'].map(mapping_dicts['discharge_disposition_id']).fillna('Not Available / Unmapped').replace('NULL', 'Not Available / Unmapped')
df_proc['admission_source_desc'] = df_proc['admission_source_id'].map(mapping_dicts['admission_source_id']).fillna('Not Available / Unmapped').replace('NULL', 'Not Available / Unmapped')

print("=== RELATIONAL ID MAPPING COMPLETED ===")
df_proc[['admission_type_id', 'admission_type_desc', 'discharge_disposition_id', 'discharge_disposition_desc']].head(5)
```

```python
zero_var_cols = [col for col in df_proc.columns if df_proc[col].nunique(dropna=False) == 1]
print(f"Zero Variance Features Identified: {zero_var_cols}")
```

```python
df_proc = df_proc.drop(columns=zero_var_cols + ['weight'])
print(f"Shape after dropping zero-variance features and weight: {df_proc.shape}")
```

## 7. Categorical Standardization
High-cardinality features and unstandardized categorical text increase feature space complexity. Here we convert age brackets to numerical midpoints and map 900+ ICD-9 diagnosis codes into 9 primary clinical categories.

```python
age_map = {
    '[0-10)': 5, '[10-20)': 15, '[20-30)': 25, '[30-40)': 35, '[40-50)': 45,
    '[50-60)': 55, '[60-70)': 65, '[70-80)': 75, '[80-90)': 85, '[90-100)': 95
}
df_proc['age_midpoint'] = df_proc['age'].map(age_map).astype(float)
```

```python
def map_icd9_category(code):
    if pd.isna(code) or code == '?' or code == 'Unknown':
        return 'Missing/Unknown'
    code_str = str(code).strip()
    if code_str.startswith('V') or code_str.startswith('E'):
        return 'Other'
    try:
        val = float(code_str)
    except ValueError:
        return 'Other'
    
    if (390 <= val <= 459) or val == 785:
        return 'Circulatory'
    elif (460 <= val <= 519) or val == 786:
        return 'Respiratory'
    elif (520 <= val <= 579) or val == 787:
        return 'Digestive'
    elif int(val) == 250:
        return 'Diabetes'
    elif 800 <= val <= 999:
        return 'Injury'
    elif 710 <= val <= 739:
        return 'Musculoskeletal'
    elif (580 <= val <= 629) or val == 788:
        return 'Genitourinary'
    elif 140 <= val <= 239:
        return 'Neoplasms'
    else:
        return 'Other'

for col in ['diag_1', 'diag_2', 'diag_3']:
    df_proc[f'{col}_category'] = df_proc[col].apply(map_icd9_category)

print("=== ICD-9 DIAGNOSIS CATEGORIZATION SAMPLE (diag_1_category) ===")
print(df_proc['diag_1_category'].value_counts())
```

## 8. Missing-Value Treatment
We apply context-specific imputation strategies grounded in clinical data science best practices.

```python
df_clean = df_proc.copy()
df_clean = df_clean.dropna(subset=['gender'])
```

```python
df_clean['race'] = df_clean['race'].fillna('Unknown')
df_clean['payer_code'] = df_clean['payer_code'].fillna('Missing')
df_clean['medical_specialty'] = df_clean['medical_specialty'].fillna('Missing')
df_clean['diag_1'] = df_clean['diag_1'].fillna('Unknown')
df_clean['diag_2'] = df_clean['diag_2'].fillna('Unknown')
df_clean['diag_3'] = df_clean['diag_3'].fillna('Unknown')

print(f"Total unhandled missing values in df_clean: {df_clean.isnull().sum().sum()}")
print(f"Final Clean Dataset Shape: {df_clean.shape[0]:,} rows x {df_clean.shape[1]} columns")
```

## 11. Assumptions & Data-Preparation Decisions
Every significant cleaning decision is formally documented under the **Problem → Decision → Reason** governance format.

| # | Problem Identified | Decision Applied | Data-Science & Clinical Rationale |
|---|--------------------|------------------|----------------------------------|
| **1** | Implicit missing value markers (`'?'`, `'Unknown/Invalid'`) in raw CSV. | Replaced tokens with explicit `np.nan` before missingness profiling. | Standardizes missingness handling; prevents implicit string placeholders from corrupting statistical aggregation or categorical encodings. |
| **2** | `weight` column is 96.85% missing (98,569 rows out of 101,766). | Dropped `weight` column completely from the dataset. | High missingness (>96%) provides insufficient coverage. Imputing >96% of records would introduce high artificial variance and bias. |
| **3** | `payer_code` (39.56% missing) & `medical_specialty` (49.08% missing). | Imputed missing values with explicit category `'Missing'`. | Missingness in administrative payer code or physician specialty represents informative missingness (e.g. self-pay/uninsured patients or unassigned specialty). Preserving missingness as a distinct category avoids discarding ~50% of hospital encounters. |
| **4** | Lab test features (`max_glu_serum` & `A1Cresult`) stored as `NaN` / `'None'`. | Explicitly encoded `'Not Tested'` as a valid clinical category. | Missing lab test results indicate that no test was ordered during the encounter. Retaining `'Not Tested'` preserves critical clinical protocol decision context and prevents CSV parsing misinterpretations. |
| **5** | Relational ID lookups contain string `'NULL'`. | Replaced `'NULL'` text with `'Not Available / Unmapped'`. | Prevents standard CSV parsers from misinterpreting string `'NULL'` labels as null values upon file read-back. |
| **6** | `race` feature missing in 2.23% of records (2,273 rows). | Imputed missing values with explicit category `'Unknown'`. | Preserves 2,273 valid clinical encounters without introducing artificial demographic bias through mode imputation. |
| **7** | `gender` contains 3 `'Unknown/Invalid'` records (0.003%). | Filtered out the 3 invalid records. | Invalid demographic entries represent a microscopic fraction (<0.003%) of data; dropping them resolves record validity without statistical impact. |
| **8** | `examide` and `citoglipton` features contain only `'No'` across all 101,766 rows. | Dropped both features from the dataset. | Zero-variance constant features contain zero entropy/information and cannot contribute predictive power to downstream analysis or machine learning models. |
| **9** | Raw ICD-9 diagnosis codes (`diag_1`, `diag_2`, `diag_3`) contain 900+ distinct code values. | Grouped ICD-9 codes into 9 high-level disease categories (Circulatory, Respiratory, Digestive, Diabetes, Injury, Musculoskeletal, Genitourinary, Neoplasms, Other). | Solves high-cardinality and data sparsity issues, reducing overfitting risks while aligning features with standard epidemiological classification literature. |
| **10** | Categorical variables `admission_type_id`, `discharge_disposition_id`, and `admission_source_id` stored as numeric integers. | Mapped integer keys to descriptive text labels using `IDS_mapping.csv`. | Prevents machine learning algorithms from incorrectly interpreting arbitrary relational ID numbers as ordinal or continuous values. |
| **11** | Patient identifier `patient_nbr` has 30,248 repeat admission records. | Retained all 101,763 encounter rows. | Each row represents a distinct clinical encounter. Longitudinal repeat admissions are essential for modeling 30-day hospital readmission outcomes. |
| **12** | Deceased patients (`discharge_disposition_id` 11, 19, 20, 21 - Expired). | Retained in prepared dataset; flagged for downstream Task 2 EDA / Task 3 filtering. | Preserves full data preparation completeness while establishing clinical constraints for subsequent readmission modeling (expired patients cannot be readmitted). |
